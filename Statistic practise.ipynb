```python
import numpy as np
import statistics as stats
import pandas as pd
```

```python
                # mean#
# with numpy
data = [50,70,40,60,90]
mean_value = np.mean(data)
mean_value
```

```python
# with stats
mean = stats.mean(data)
mean
```

```python
                       # median #
# with numpy
data1 = [51,45,101,36,28,8,0]
median_value = np.median(data1)
median_value
```

```python
# with stats
median = stats.median(data1)
median
```

```python
                      # mode #
# with stats
data2 = [3,4,2,1,3,2,4,3,4,1,3]
mode_value = stats.mode(data2)
mode_value
```

```python
data3 = [5,1,9,9,5,2,5,2,1,5,2,9,2]
mode = stats.mode(data3)   # => we can use multimode() function
mode
```

```python
''' Q1. in a math quiz ,there are scores of 11 students:
[45,25,48,52,60,58,52,49,55,60,63]
a) Find the mean of the scores.
b) find the median scores.
c) Find the mode of the scores '''
```

```python
DATA = [45,52,48,52,60,58,52,49,55,60,63]

# mean
ans1 = stats.mean(DATA)
print(ans1)

# median
ans2 = stats.median(DATA)
print(ans2)

# mode
ans3 = stats.mode(DATA)
print(ans3)
```
