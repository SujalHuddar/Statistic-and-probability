```python
# outliers
age = [18,25,30,45,33,67,78,93,250,76,99,271]
age
```

```python
import matplotlib.pyplot as plt
```

```python
plt.boxplot(age)
plt.show()
```

```python
# to remove outlires
''' steps
1. calculare Q1 and Q3
2. Find IQR
3. Find the upper and lower limit
4.  Take only that value which which will be less than or equal to my upper limit and grater 
    than or equal to my lower limit'''
```

```python
# step-1: calculare Q1 and Q3
q1 = np.percentile(age,25)
q1
```

```python
q3 = np.percentile(age, 75)
q3
```

```python
# step-2: find IQR
IQR = q3-q1
IQR
```

```python
# step-3: Find the upper and lower limit
ll = q1-1.5*IQR
ul = q3+1.5*IQR
print(ll)
print(ul)
```

```python
# step-4: Take only that value which which will be less than or equal to my upper limit and grater 
#    than or equal to my lower limit
for i in age:
    if i>ul or i<ll:
        print(f'Outlier values in the data is {i} and it is being deteceated')
```

```python
clean_age = []
for i in age:
    if i>ul or i<ll:
        print(f'Outlier values in the data is {i} and it is being deteceated')
    else :
        clean_age.append(i)
```

```python
clean_age
```

```python
plt.boxplot(clean_age)
plt.show()
```

```python
'''Question:
data = [36,38,40,42,45,46,47,48,49,100,160,-10,-20]
a. show the data using boxplot
b. find the q1, q2 and q3
c. find IQR and ul and ll
d. identify the outliers and if there is any outliers remove them
e. plot the data using the boxplot without outliers'''
```

```python
data = [36,38,40,42,45,46,47,48,49,100,160,-10,-20]
data
```

```python
plt.boxplot(data)
plt.show()
```

```python
Q1 = np.percentile(data,25)
Q2 = np.percentile(data,50)
Q3 = np.percentile(data,75)
print(Q1)
print(Q2)
print(Q3)
```

```python
iqr = Q3-Q1
iqr
```

```python
ll = Q1-1.5*iqr
ul = Q3+1.5*iqr
print(ll)
print(ul)
```

```python
clean_data = []
for i in data:
    if i>ul or i<ll:
        print(f"outliers in the data is {i} and is being detected")
    else:
        clean_data.append(i)
```

```python
print(clean_data)
```

```python
plt.boxplot(clean_data)
plt.show()
```
